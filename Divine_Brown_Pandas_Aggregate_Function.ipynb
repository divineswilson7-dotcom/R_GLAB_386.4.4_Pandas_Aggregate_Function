```python
# R-GLAB 386.4.4 - Pandas Aggregate Function Lab

import pandas as pd

# Create the DataFrame
data = {
    'Department': [
        'Sales', 'Sales', 'Sales', 'Sales', 'Sales', 'Sales',
        'Engineering', 'Engineering', 'Engineering', 'Engineering', 'Engineering',
        'HR', 'HR', 'HR', 'HR',
        'Marketing', 'Marketing', 'Marketing', 'Marketing', 'Marketing'
    ],

    'Employee': [
        'Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank',
        'Grace', 'Hannah', 'Ian', 'Jack', 'Karen',
        'Liam', 'Mona', 'Nathan', 'Olivia',
        'Paul', 'Quinn', 'Rachel', 'Sam', 'Tina'
    ],

    'Salary': [
        70000, 80000, 75000, 82000, 68000, 79000,
        110000, 120000, 115000, 105000, 125000,
        65000, 60000, 62000, 67000,
        72000, 78000, 85000, 69000, 74000
    ],

    'Bonus': [
        5000, 8000, 6000, 8500, 4500, 7500,
        12000, 15000, 13000, 11000, 16000,
        4000, 3000, 3500, 4200,
        5500, 7000, 9000, 5000, 6000
    ],

    'Performance_Score': [
        88, 92, 85, 90, 82, 89,
        95, 98, 94, 91, 99,
        78, 82, 80, 84,
        87, 91, 93, 83, 88
    ]
}

df = pd.DataFrame(data)

df.head(20)
```

```python
# Task 1: Calculate total, average, max values for salary in one command

df[['Salary']].agg(['sum', 'mean', 'max'])
```

```python
# Task 2: Calculate min, max and average across Salary and Bonus columns

df[['Salary', 'Bonus']].agg(['min', 'max', 'mean'])
```

```python
# Task 3: Calculate specific metrics for each column

df.groupby('Department').agg(
    {
        'Salary': ['mean', 'median'],
        'Performance_Score': ['max'],
        'Bonus': ['sum']
    }
)
```

```python
# Task 4: For each department - average salary,
# highest performance score, total bonus paid

df.groupby('Department').agg(
    {
        'Salary': ['mean'],
        'Performance_Score': ['max'],
        'Bonus': ['sum']
    }
)
```
